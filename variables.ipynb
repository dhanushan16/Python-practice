```python
#variable-declaration
#syntax=variable_name=value

var1=50
print(var1)

var2="hi how are you"
print(var2)

var3=3.97
print(var3)
```

```python
#variable-redeclaration

num=5 #declaring
print(num)

num=6 #redeclaring
print(num)

num="hi" #redeclaring
print(num)
```

```python
#assigning single value to multiple variables

num1=num2=num3=25
print(num1)
print(num2)
print(num3)
```

```python
#multiple assignments

a,b,c="hi",3,093.342
print(a)
print(b)
print(c)
```
