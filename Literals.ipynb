```python
#string literals
str1='welcome'
print (str1)

str2="welcome"
print(str2)

str3='''welcome all'''
print(str3)

str4="""welcome all"""
print(str4)

str5='Hi\
hello\
good morning'
print(str5)
```

```python
#character literal

var1='x'
print(var1)

var2="y"
print(var2)
```

```python
#numeric literals
#decimal-number without fraction

a=213
print(a)

b=-3183
print(b)

#binary-starts with 0b followed by 0 and 1

c=0b1111
print(c)

d=0b000111
print(d)

#octal-starts with 0o 
e=0o123
print(e)

#hexadecimal-starts with 0x
f=0x24121
print(f)

#floating point- number + fraction

g=0.29412
print(g)

#complex- real+ imaginery

h=2+8j
print(h)
```

```python
#boolean literals- true=1 false=0
a=(1==1)
print(a)

b=(1==False)
print(b)

c=True+5
print(c)

d=False+5
print(d)
```

```python
#special literal-none used to indicate absence of value or no value

value=None
print(value)
```
