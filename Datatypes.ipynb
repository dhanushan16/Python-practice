```python
#numeric datatypes

#integer
i=3
print(i)
print(type(i))
print(id(i))

#float 
f=0.385
print(f)
print(type(f))

f1=float(4)
print(f1)
print(type(f1))

#complex

x=3+4j
y=6.4-3.5j
print(x)
print(y)
print(type(x))
print(type(y))
print(x.real)#accessing real part
print(x.imag)#accessing imaginery part
print(y.real)#accessing real part
print(y.imag)#accessing imaginery part
```

```python
#boolean data type

a=29
b=34
result=a>b
print(result)#false
print(type(result))
output=a<b
print(output)#true
print(type(output))
```

```python
#sequence data type

#string  data type
str1="welcome all"
print(str1)
print(type(str1))

#accesing str elements through array index number starting from 0
print(str1[0])      # Prints the first character
print(str1[1])      # Prints the second character
print(str1[-1])     # Prints the last character
print(str1[6:11])   # Prints characters from index 6 to 10
print(str1[2:])     # Prints from index 2 to the end
print(str1[:5])     # Prints from the beginning up to index 4
print(str1[:-9])    # Prints from the beginning up to 9 characters before the end
print(str1[-9:])    # Prints the last 9 characters

#list data type

list=[1,'a',"abc",3.281]
print(list)
print(list[0])
print(list[1])
print(list[2])
print(list[3])
print(type(list))
print(type(list[0]))

list1=[2,'dj',[2,5,7],"elephant"]#list inside list
print(list1[2])
print(list1[2][0])

#tuple data type

t=('hi',3038,3.45)
print(t)
print(t[0])
print(type(t))
print(type(t[0]))
```

```python
#mapping

#dictionary
dict={'karnataka':'bengaluru','tamil nadu':'chennai'}
print(dict)
print(dict['tamil nadu'])
print(type(dict))
```

```python
#sets

#set
intset={1,2,3,4,6,6,6,9}
print(intset)
student={21,'ramu',99.99}
print(student)
print(type(student))

#frozenset
student=frozenset({21,'ramu',99.99})
print(student)
print(type(student))
```

```python
#byte
x = bytes([10, 20, 30, 40])

print(x)
print(x[0])

name = "Python"
x = name.encode()
print(x)
print(x[0])

#bytearray

x = bytearray([10, 20, 30, 40])

print(x)
print(x[0])
x[0]=38
print(x)
print(x[0])#mutable
```
