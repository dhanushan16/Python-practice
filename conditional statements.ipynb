```python
#calculating bonus based on salary (if else)

salary = float(input("Enter your salary: "))

if salary >= 50000:
    bonus = salary * 0.10
else:
    bonus = salary * 0.05

total_salary = salary + bonus

print("Bonus:", bonus)
print("Total Salary:", total_salary)
```

```python
#program to check ticket pricing based on age(elif)

age=int(input("enter your age to check ticket pricing:"))
if age<5:
    print("ticket is free!")
elif age<=12:
    print("child ticket:10rs")
elif age<=60:
    print("adult ticket:30rs")
else:
    print("senior citizen:20rs")
```

```python
#program to find out weather a number is 0 ,+ve or -ve(nested if)

number=int(input("enter your number:"))
if number>=0:
    if number==0:
        print("number is zero")
    else:
        print("number is positive")
else:
    print("number is negative")
    
```

```python
#program to find which is the largest number of 3 numbers(nested if)

n1=int(input("enter first number:"))
n2=int(input("enter second number:"))
n3=int(input("enter third number:"))
if n1>n2:
    if n1>n3:
        print(f"{n1} is greater")
    else:
        print(f"{n3} is  greater")
else:
    if n2>n3:
        print(f"{n2} is greater")
    else:
        print(f"{n3} is greater")
        
```
