```python
#arithmetic operators

x=10
y=5
print(x+y) #addition
print(x-y) #subtraction
print(x*y) #multiplication
print(x/y) #division
print(x%y) #modulas
print(x//y) #floor division
print(x**y) #exponential
```

```python
#assignment operators

a=16#equals
print(a)

a+=8 #a=a+8
print(a)

a-=8 #a=a-8
print(a)

a*=8 #a=a*8
print(a)

a/=8 #a=a/8
print(a)

a%=8 #a=a%8
print(a)

a//=8 #a=a//8
print(a)

a**=8 #a=a**8
print(a)

```

```python
#comparision or relational operators 

x=30
y=20
print(x==y)
print(x!=y)
print(x>y)
print(x>=y)
print(y<x)
print(y<=x)

a=10
b=10
print(a==b)
print(a!=b)
print(a>b)
print(a>=b)
print(a<b)
print(a<=b)
```

```python
#logical operators(and, or , not)

#and
x=2
y=3
print((x>1) and(y>=3))

#or
a=6
b=8
print((a>4) or (b<5))

#not
print(not True)
print(not False)
```

```python
#membership operators(in, not in)

x=[5,6,7]
print(6 in x)
print(9 in x)
print(9 not in x)
print(6 not in x)
```

```python
#identity operators( is, is not)

x=["hi","hello"]
y=["hi","hello"]
z=x

print(x is z) 
print(x is y)
print(x==y)

print(x is not z)
print(x is not y)
print(x!=z)
```

```python
#binary operators
a=3
b=6
print(a & b)
print(a | b)
print( ~ a)
x=0b1010
print(x<<3)
print(x>>2)
```
